## Bài tập thực hành
### Bài 1: Tạo tensor cơ bản
Tạo một Tensor số nguyên kích thước 2 × 3 và hiển thị kết quả.


```python
a = torch.tensor([[1, 2, 3], [4, 5, 6]], dtype=torch.float32)
print(a)

#Kích thước, số chiều, số lượng phần tử
print(f'Kích thước: {a.shape}')
print(f'Số chiều: {a.ndim}')
print(f'Kiểu dữ liệu: {a.dtype}')
```

### Bài 2. Tạo các Tensor thường dùng
Thực hành tạo Tensor ngẫu nhiên, Tensor toàn số 0, toàn số 1 và ma trận đơn vị.

```python
a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)
print(a)
print(b)
print(c)
print(d)
print(e)
```

**Nhận xét**:
| `torch.rand()` | `torch.randn()` |
|-----|------|
| Ngẫu nhiên theo phân phối đều | Ngẫu nhiên theo phân phối chuẩn |
| Các giá trị trong `[0,1)` có xác suất như nhau | Giá trị tập trung quanh `0`, xa `0` thì xác suất càng loãng|

```python
# Tạo một tensor 5 x 5 với tất cả các phần tử bằng 10
f = torch.full((5, 5), 10)
print(f)
```

### Bài 3. Kiểm tra thông tin Tensor
Sử dụng shape, size(), ndim và dtype để kiểm tra các thuộc tính của Tensor.


```python
a = torch.zeros(2, 3, 4)

print("Tensor:")
print(a)
print("Shape:", a.shape)
print("Size:", a.size())
print("Number of dimensions:", a.ndim)
print("Data type:", a.dtype)
```

Ý nghĩa của **`torch.Size([2,3,4])`**: cho thấy kích thước của các chiều, mô tả cách dữ liệu được tổ chức: gồm 2 khối chứa 3 hàng, mỗi hàng chứa 4 phần tử

### Bài 4. Slicing và Indexing
Truy xuất một phần dữ liệu từ Tensor nhiều chiều.

```python
import torch
a = torch.randn(3, 4, 5)
print("Original tensor:")
print(a)
print("\nTensor a[1]:")
print(a[1])
print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])
```

- Shape của `a[1]` là kích thước của dim0(chiều 1) lấy index 1 `torch.Size([4,5])`
- Shape của `a[1:, 2:4]` là kích thước của dim0(chiều 1) lấy index từ 1 và dim1(chiều 2) lấy index 2,3 `torchSize([2,2,5])`
- Cách lấy 2 phần tử cuối cùng trên chiều thứ 3(dim2) là `a[:,:,-2:]`


```python
print("Shape a[1]:", a[1].shape)
print("Shape a[1:, 2:4]:", a[1:, 2:4].shape)
print(a[:,:,-2:])
```

### Bài 5. Reshape và Flatten
Thay đổi hình dạng Tensor mà không làm thay đổi số phần tử.

```python
a = torch.randn(3, 4)
print("Original:")
print(a)
print("\nFlatten:")
print(a.ravel())
print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))
```
