### Bài 8. Autograd cơ bản
Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.

```python
x = torch.tensor(3.6, requires_grad=True)

y = x**2 
y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad) # Output: tensor(7.2000)
```

**Tính thủ công và so sánh đáp án của $y= x^2$ tại $x=3.6$**
$$ y = x^2$$
$$\implies y' = 2x$$
$$\implies y'(3.6) = 2*3.6 = 7.2$$

Đáp án đúng như `x.grad` đã tính

```python
x = torch.tensor(3.6, requires_grad=True)

y = 3*x**2 + 2*x + 1
# y' = 6*x + 2 => y'(3.6) = 6*3.6 + 2 = 21.6 + 2 = 23.6
y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad) # Output: tensor(23.6000)
```

### Bài 9. Gradient Descent tìm hệ số đa thức
Sinh dữ liệu từ đa thức $y = x^2 + 2x + 3$, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
polynomial = np.poly1d([1, 2, 3])  # x^2 + 2x + 3
N = 50

X = np.random.randn(N, 1) * 5
Y = polynomial(X)
Y += np.random.randn(N, 1) * 2   # Thêm nhiễu Gaussian vào dữ liệu

XX = np.hstack([X*X, X, np.ones_like(X)])  # Tạo feature matrix với các cột là x^2, x và 1
# y = w0*x^2 + w1*x + w2...
w = torch.randn(3,1,requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01) 

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    MSE = torch.mean((y - y_pred) ** 2)
    MSE.backward()
    optimizer.step()

print("Learned coefficients weights:", w)   
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, Y, alpha=0.5, label="Data")
X_plot = np.linspace(X.min(), X.max(), 300).reshape(-1, 1)

XX_plot = np.hstack([
    X_plot**2,
    X_plot,
    np.ones_like(X_plot)
])

x_plot = torch.tensor(XX_plot, dtype=torch.float32)

with torch.no_grad():
    Y_pred = (x_plot @ w).numpy()
# Phương trình model học được
equation = "y = {:.2f}x² + {:.2f}x + {:.2f}".format(
    w[0].item(),
    w[1].item(),
    w[2].item()
)

plt.plot(
    X_plot,
    Y_pred,
    label=f"Learned: {equation}"
)

plt.plot(
    X_plot,
    polynomial(X_plot),
    "--",
    label="Reality: y = x² + 2x + 3"
)

plt.xlabel("X")
plt.xlim(-10, 10)
plt.ylim(-5, 100)
plt.ylabel("Y")
plt.legend()
plt.grid()
plt.show()
```

**Nhận xét:**

- Trọng số `w` sau khi học được đã gần xấp xỉ được hệ số ban đầu`[1, 2, 3]`

- ``optimizer.zero_grad()``: xóa gradient của vòng lặp trước đó, nếu không gradient cũ sẽ bị cộng dồn khi gọi `mse.backward()` (cơ chế của pytorch).
- ``mse.backward()``: tính gradient của bình phương sai số(mse) để thực hiện tìm hướng thay đổi `w` để giảm `mse`.
- ``optimizer.step()``: dùng gradient đã tính trước đó để cập nhật trọng số.

### Bài 10. Giải hệ phương trình bằng Autograd
Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

```python
import random

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
   y1 = A + B - 9
   y2 = C - D - 1
   y3 = A + C - 8
   y4 = B - D - 2

   SE = y1**2 + y2**2 + y3**2 + y4**2

   optimizer.zero_grad()
   SE.backward()
   optimizer.step()

print("A = " ,A)
print("B = " ,B)
print("C = " ,C)
print("D = " ,D)
```

**Kiểm tra nghiệm thu**
$$
\begin{cases}
y_1 = A + B - 9 \\
y_2 = C - D - 1 \\
y_3 = A + C - 8 \\
y_4 = B - D - 2
\end{cases}
$$

Mục tiêu là tìm $A,B,C,D$ sao cho:

$$
y_1 = y_2 = y_3 = y_4 = 0
$$

Hay tương đương:

$$
\begin{cases}
A + B = 9 \\
C - D = 1 \\
A + C = 8 \\
B - D = 2
\end{cases}
$$
Kết luận phương trình có ***vô số nghiệm***

Ta thu được hệ nghiệm tổng quát:
$$A=t, B=9−t, C=8−t, D=7−t​$$

```python
#Kiểm thử sai số:
import numpy as np

A = A.item()
B = B.item()
C = C.item()
D = D.item()

# Chọn A làm mốc
AA = A
BB = 9 - AA
CC = 8 - AA
DD = 7 - AA

print(f'Sai số biến A:{AA - A}')
print(f'Sai số biến B:{BB - B}')
print(f'Sai số biến C:{CC - C}')
print(f'Sai số biến D:{DD - D}')
```

**Nhận xét:**
- Sai số xấp xỉ bằng 0 cho thấy nghiệm tìm bằng Gradient Descent phù hợp với nghiệm lý thuyết.
- Bài toán có vô số nghiệm

```python
# y = 2x² - 3x + 5
poly = np.poly1d([2, -3, 5])

# Data
X = np.random.randn(50, 1) * 5
Y = poly(X)

# Features: [x², x, 1]
XX = np.hstack([X**2, X, np.ones_like(X)])

x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

w = torch.randn(3, 1, requires_grad=True)

optimizer = torch.optim.NAdam([w], lr=0.01)

for _ in range(1000):
    optimizer.zero_grad()

    y_pred = x @ w
    loss = ((y - y_pred) ** 2).mean()

    loss.backward()
    optimizer.step()

print("Learned w:", w.detach().flatten())
```
