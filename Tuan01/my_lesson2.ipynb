## ex1: Tính giá trị hàm số

#### Nội dung bài
Hãy viết chương trình nhập vào số thực $x$ và tính giá trị biểu thức $f(x) = x + \frac{x^5}{5!} + \frac{\sqrt{|x|}}{x^\frac{3}{2}}$

#### Bài làm

```python
import math

x = float(input('Moi ban nhap vao gia tri cua bien so x: '))
#x = float(input('Moi ban nhap vao gia tri cua bien so x: '))
print("Ban da nhap: ", x)
if (x==0): print('Không xác định')
else:
    a = math.factorial(5)
    b = math.sqrt(abs(x))
    f_x = x + (x**5)/a + b/(x**(3/2))
print(f'Gia tri cua ham so f({x}) = {f_x: .2f}.')
```

## ex7: Giải phương trình bậc 2

### Nội dung bài: 
Nhập 3 số thực $a$, $b$, $c$ (có kiểm tra dữ liệu nhập số thực). Hãy xây dựng hàm giải phương trình bậc 2 là
$ax^2+bx+c=0$ và in kết quả giải ra màn hình.

### Bài làm

#### Nhập dữ liệu

```python
def NhapDuLieu():
    """
    Input:
    Ban phim = 1 5 6
    Output:
    (a, b, c) --> (1, 5, 6)
    """
    
    a = float(input('Nhap he so a: '))
    b = float(input('Nhap he so b: '))
    c = float(input('Nhap he so c: '))

    # ...
    return a, b, c

a, b, c = NhapDuLieu()
print(f'a = {a}, b = {b}, c = {c}')
```

#### Hàm giải phương trình bậc 2

```python
def GiaiPhuongTrinhBac2(a, b, c):
    """
    Input: a, b, c
    Output:
    + flag = -1 (VSN), 0 (VN), k (k nghiem)
    + () --> flag = -1, 0
    + (x) --> flag = 1
    + (x1, x2) --> flag = 2
    """
    import math
    flag = None
    x = ()
    
    if a==0: 
        if b==0:
            if c==0: 
                flag = -1
                x = ()
            else:
                flag = 0
                x = ()
        else:
            flag = 1
            x = (-c/b,)
    else:
        delta = b*b - 4*a*c
        if delta < 0:
            flag = 0
            x = ()
        elif delta > 0 :
            t = math.sqrt(delta)
            flag = 2
            x = ((-b - t)/(2*a), (-b + t)/(2*a))
        else:
            flag = 1
            x = (-b/(2*a),)

    return flag, x

# GiaiPhuongTrinhBac2
print("a = 0")
flag, x = GiaiPhuongTrinhBac2(0, 0, 0)
print(f'So nghiem: {flag}, nghiem = {x}')
flag, x = GiaiPhuongTrinhBac2(0, 0, 5)
print(f'So nghiem: {flag}, nghiem = {x}')
flag, x = GiaiPhuongTrinhBac2(0, 5, 6)
print(f'So nghiem: {flag}, nghiem = {x}')
print("a <> 0")
flag, x = GiaiPhuongTrinhBac2(5, 1, 2)
print(f'So nghiem: {flag}, nghiem = {x}')
flag, x = GiaiPhuongTrinhBac2(1, 2, 1)
print(f'So nghiem: {flag}, nghiem = {x}')
flag, x = GiaiPhuongTrinhBac2(1, 5, 6)
print(f'So nghiem: {flag}, nghiem = {x}')

"""
a = 0
So nghiem: -1, nghiem = ()
So nghiem: 0, nghiem = ()
So nghiem: 1, nghiem = (-1.2,)
a <> 0
So nghiem: 0, nghiem = ()
So nghiem: 1, nghiem = -1.0
So nghiem: 2, nghiem = (-3.0, -2.0)
"""
```

#### Kết nối bài

```python
# Kết nối bài
a, b, c = NhapDuLieu()
# Gọi hàm giải
flag, x = GiaiPhuongTrinhBac2(a, b, c)
# Xuất dữ liệu
s = f'Phuong trinh bac 2 {a}x^2 + {b}x + {c} = 0'
if flag == -1:
    print(f'{s} co vo so nghiem!')
elif flag == 0:
    print(f'{s} vo nghiem!')
elif flag == 1:
    print(f'{s} co 1 nghiem, x = {x[0]}!')
elif flag == 2:
    print(f'{s} co 2 nghiem, x1 = {x[0]}, x2 = {x[1]}!')
# if
# Moi ban nhap he so a, b, c: 1 5 6
# Phuong trinh bac 2 1x^2 + 5x + 6 = 0 co 2 nghiem, x1 = -3.0, x2 = -2.0!
```

Viết chương trình xây dựng các hàm sau cho List và kiểm tra tính đúng đắn của các hàm số:
+ Nhập một mảng số từ bàn phím cách nhau bởi khoảng trắng. Xuất mảng ra màn hình.
+ Sinh một mảng số ngẫu nhiên có n phần tử trong khoảng a, b.
+ Tính tổng của mảng số, đếm các phần tử chẵn, lẻ trong mảng.
+ Cho một mảng số, hãy trả về dãy chẳn, dãy lẻ.

### Bài làm

#### Nhập mảng

```python
def NhapMang():
    """
    Input:
    Ban phim = 1 4 1 2 3 4 1 2 3 3 1 2 3 4 10 2
    Output:
    a = [1, 4, 1, 2, 3, 4, 1, 2, 3, 3, 1, 2, 3, 4, 10, 2]
    """
    a = list(map(int, input("Moi ban nhap mang (cach nhau khoang trang): ").split()))
    # ...
    return a

# NhapMang
a = NhapMang() # 1 4 1 2 3 4 1 2 3 3 1 2 3 4 10 2
print(a)
# Moi ban nhap mang (cach nhau khoang trang): 1 4 1 2 3 4 1 2 3 3 1 2 3 4 10 2
# ['1', '4', '1', '2', '3', '4', '1', '2', '3', '3', '1', '2', '3', '4', '10', '2']
```

#### Sinh ngẫu nhiên

```python
import random

def SinhNgauNhien(n, vmin = -10, vmax = 10):
    """
    Input:
    n = 30, vmin = -10, vmax = 10
    Output:
    a = [-1, -4, -8, 6, 4, -5, -1, -4, -9, -5, -10, 0, -6, 2, 1, -9, -9, 10, -8, -9, -7,
    9, 2, -4, -1, -6, 5, -6, 9, -8]
    """
    a = []
    
    for i in range(1, n):
        a.append(random.randint(vmin, vmax))
    # ...
    return a

# SinhNgauNhien
b = SinhNgauNhien(30, -10, 10)
print(b)
b = SinhNgauNhien(30, vmin = 0)
print(b)
b = SinhNgauNhien(30, vmax = 2)
print(b)
```

#### Đếm tổng chẵn lẻ

```python
def DemTongChanLe(a):
    """
    Input: a[] = [1, 4, 2, 0, -1, -7, 3, -6, 4, 9, -5, 7, 7, -7, -10, 1, -10, 9, -4, -6, 10, -
    2, -10, -8, -3, 2, -7, -10, 6, 0]
    Output: tong, sochan, sole
    """
    tong = 0
    sochan = 0
    sole = 0
    
    for i in a:
        tong += i
        if not i%2: sochan += 1
        else: sole += 1

    # ...
    return tong, sochan, sole

# DemTongChanLe
a = [1, 4, 2, 0, -1, -7, 3, -6, 4, 9, -5, 7, 7, -7, -10, 1, -10, 9, -4, -6, 10, -2, -10, -8, -
3, 2, -7, -10, 6, 0]
print(DemTongChanLe(a))
a = [9, 6, 5, 0, 8, 8, 1, 10, 0, 0, 1, 1, 2, 7, 1, 10, 9, 1, 10, 8, 2, 8, 4, 5, 4, 5, 5, 3, 6,
2]
print(DemTongChanLe(a))
# (-31, 17, 13)
# (141, 17, 13)
```

#### Xuất chẵn lẻ

```python
def DayChanLe(a): # Gia tri tra ve
    """
    Input:
    a = [1, 4, 2, 0, -1, -7, 3, -6, 4, 9, -5, 7, 7, -7, -10, 1, -10, 9, -4, -6, 10, -2, -
    10, -8, -3, 2, -7, -10, 6, 0]
    Output: (achan, ale)
    achan = [4, 2, 0, -6, 4, -10, -10, -4, -6, 10, -2, -10, -8, 2, -10, 6, 0]
    ale = [1, -1, -7, 3, 9, -5, 7, 7, -7, 1, 9, -3, -7]
    """
    achan = []
    ale = []
    
    for i in a:
        if not i%2: achan.append(i)
        else: ale.append(i)

    # ...
    return achan, ale

# DayChanLe
a = [1, 4, 2, 0, -1, -7, 3, -6, 4, 9, -5, 7, 7, -7, -10, 1, -10, 9, -4, -6, 10, -2, -10, -8, -
3, 2, -7, -10, 6, 0]
achan, ale = DayChanLe(a)
print(f'DayChanLe:\nChan = {achan}\nLe = {ale}\n')
# DayChanLe:
# Chan = [4, 2, 0, -6, 4, -10, -10, -4, -6, 10, -2, -10, -8, 2, -10, 6, 0]
# Le = [1, -1, -7, 3, 9, -5, 7, 7, -7, 1, 9, -3, -7]
```

## ex9: Các thao tác trên tập hợp, từ điển

### Nội dung bài:
Viết chương trình nhập vào một dãy ký tự a cách nhau bởi khoảng trắng. Hãy thực hiện:
+ In các ký tự trong a không trùng nhau và được sắp theo thứ tự tăng dần
+ Đếm số lần xuất hiện các ký tự trong dãy a

### Bài làm

#### Xử lý không trùng

```python
def XuLyKhongTrung(a):
    """
    Input:
    a = ['A', 'G', 'C', 'A', 'C', 'F', 'F', 'T', 'H']
    Output:
    s = ['A', 'C', 'F', 'G', 'H', 'T']
    """
    s = set([])
    for i in a:
        s.add(i)
    # ...
    return sorted(s)

# XuLyKhongTrung
a = ['A', 'G', 'C', 'A', 'C', 'F', 'F', 'T', 'H']
b = XuLyKhongTrung(a)
print(b)
# ['A', 'C', 'F', 'G', 'H', 'T']
```

#### Đếm tần số xuất hiện

```python
def DemSoLanXuatHien(a):
    """
    Input:
    a = ['A', 'G', 'C', 'A', 'C', 'F', 'F', 'T', 'H']
    Output:
    dem = {'A': 2, 'G': 1, 'C': 2, 'F': 2, 'T': 1, 'H': 1}
    """
    dem = {} # (k, v): k (ky tu) v (so lan xuat hien)
    
    for i in a:
        dem[i] = dem.get(i,0) + 1
    # ...
    return dem

# DemSoLanXuatHien
a = ['A', 'G', 'C', 'A', 'C', 'F', 'F', 'T', 'H']
b = DemSoLanXuatHien(a)
print(b)
# {'A': 2, 'G': 1, 'C': 2, 'F': 2, 'T': 1, 'H': 1}
```
