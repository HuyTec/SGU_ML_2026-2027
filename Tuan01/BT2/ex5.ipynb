# ex5_Tính tổng các số chẵn

## Nội dung bài
Nhập một số nguyên n. Hãy tính tổng các số chẵn từ 1 đến n.

## Bài làm

```python
n = int(input("Moi ban nhap so nguyen n: "))
s = 0

print("Ban da nhap:", n)

for i in range(1, n+1):
    if i % 2 == 0: s += i

print(f'Tong cac so chan tu 1 den {n} la {s}.')
```
