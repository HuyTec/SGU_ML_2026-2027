# ex1: Tính giá trị hàm số

## Nội dung bài
Hãy viết chương trình nhập vào số thực $x$ và tính giá trị biểu thức $f(x) = x + \frac{x^5}{5!} + \frac{\sqrt{|x|}}{x^\frac{3}{2}}$

## Bài làm

```python
import math

x = float(input('Moi ban nhap vao gia tri cua bien so x: '))
%run ex1_input.py
#x = float(input('Moi ban nhap vao gia tri cua bien so x: '))
print("Ban da nhap: ", x)
if (x==0): print('Không xác định')
else:
    a = math.factorial(5)
    b = math.sqrt(abs(x))
    f_x = x + (x**5)/a + b/(x**(3/2))
print(f'Gia tri cua ham so f({x}) = {f_x: .2f}.')
```
