# ex7: Giải phương trình bậc 2

## Nôi dung bài:
Nhập 3 số thực $a$, $b$, $c$ (có kiểm tra dữ liệu nhập số thực). Hãy xây dựng hàm giải phương trình bậc 2 là
$ax^2+bx+c=0$ và in kết quả giải ra màn hình.

## Bài làm

### Hàm nhập dữ liệu

```python
def NhapDuLieu():
    """
    Input:
    Ban phim = 1 5 6
    Output:
    (a, b, c) --> (1, 5, 6)
    """
    
    a = float(input('Nhap he so a: '))
    b = float(input('Nhap he so b: '))
    c = float(input('Nhap he so c: '))

    # ...
    return a, b, c

    a, b, c = NhapDuLieu()
 
```

### Hàm giải phương trình bậc 2

```python
def GiaiPhuongTrinhBac2(a, b, c):
    """
    Input: a, b, c
    Output:
    + flag = -1 (VSN), 0 (VN), k (k nghiem)
    + () --> flag = -1, 0
    + (x) --> flag = 1
    + (x1, x2) --> flag = 2
    """
    import math
    flag = None
    x = ()
    
    if a==0: 
        if b==0:
            if c==0: 
                flag = -1
                x = ()
            else:
                flag = 0
                x = ()
        else:
            flag = 1
            x = (-c/b,)
    else:
        delta = b*b - 4*a*c
        if delta < 0:
            flag = 0
            x = ()
        elif delta > 0 :
            t = math.sqrt(delta)
            flag = 2
            x = ((-b - t)/(2*a), (-b + t)/(2*a))
        else:
            flag = 1
            x = (-b/(2*a),)

    return flag, x

# GiaiPhuongTrinhBac2
print("a = 0")
flag, x = GiaiPhuongTrinhBac2(0, 0, 0)
print(f'So nghiem: {flag}, nghiem = {x}')
flag, x = GiaiPhuongTrinhBac2(0, 0, 5)
print(f'So nghiem: {flag}, nghiem = {x}')
flag, x = GiaiPhuongTrinhBac2(0, 5, 6)
print(f'So nghiem: {flag}, nghiem = {x}')
print("a <> 0")
flag, x = GiaiPhuongTrinhBac2(5, 1, 2)
print(f'So nghiem: {flag}, nghiem = {x}')
flag, x = GiaiPhuongTrinhBac2(1, 2, 1)
print(f'So nghiem: {flag}, nghiem = {x}')
flag, x = GiaiPhuongTrinhBac2(1, 5, 6)
print(f'So nghiem: {flag}, nghiem = {x}')

"""
a = 0
So nghiem: -1, nghiem = ()
So nghiem: 0, nghiem = ()
So nghiem: 1, nghiem = (-1.2,)
a <> 0
So nghiem: 0, nghiem = ()
So nghiem: 1, nghiem = -1.0
So nghiem: 2, nghiem = (-3.0, -2.0)
"""
```

### Kết nối bài

```python
# Kết nối bài
a, b, c = NhapDuLieu()
# Gọi hàm giải
flag, x = GiaiPhuongTrinhBac2(1, 5, 6)
# Xuất dữ liệu
s = f'Phuong trinh bac 2 {a}x^2 + {b}x + {c} = 0'
if flag == -1:
    print(f'{s} co vo so nghiem!')
elif flag == 0:
    print(f'{s} vo nghiem!')
elif flag == 1:
    print(f'{s} co 1 nghiem, x = {x[0]}!')
elif flag == 2:
    print(f'{s} co 2 nghiem, x1 = {x[0]}, x2 = {x[1]}!')
# if
# Moi ban nhap he so a, b, c: 1 5 6
# Phuong trinh bac 2 1x^2 + 5x + 6 = 0 co 2 nghiem, x1 = -3.0, x2 = -2.0!
```
