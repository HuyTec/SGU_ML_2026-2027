## Đọc dữ liệu

## 1. Tìm đường đi từ ô (1,1) đến ô (6,6) của mê cung bằng BFS

````python
import pprint
from ex2_maze import find_path_BFS

#..
parent = find_path_BFS(maze, wall_list, 1, 1, 6, 6)

pprint.pprint(parent)
````

### Truy vết đường đi

````python
import pprint
from ex2_maze import path

result = path(parent, 1,1,6,6)

pprint.pprint(result)
````

### Vẽ lại đường đi

````python
import matplotlib.pyplot as plt 


def draw_path(result):


    for i in range (7): ## vẽ lưới
        plt.plot([i,i],[0,6])
        plt.plot([0,6],[i,i])
    plt.gca().invert_yaxis()
    plt.axis("off")
    for wall in wall_list: ## vẽ tường
        r1, c1 = wall.a
        r2, c2 = wall.b

        r1 -= 1; c1 -= 1 ## đưa về base-0
        r2 -= 1; c2 -= 1
        
        if r1 == r2:
            r = r1
            c = max(c1,c2)
            plt.plot([c,c], [r, r+1], linewidth = 5, color = "black")
        elif c1 == c2:
            c = c1
            r = max(r1,r2)
            plt.plot([c,c+1], [r, r], linewidth = 5, color = "black")

    for point in result:
        r, c = point
        r -= 1; c-=1
        plt.fill( ## tô màu
            [c, c+1, c+1, c],
            [r, r, r+1, r+1],
            color='lightblue'
        )
    plt.fill([5, 6, 6, 5],[5, 5, 6, 6],color='red') ##tô ô end
    plt.fill([0, 1, 1, 0],[0, 0, 1, 1],color='lightgreen') ## tô ô start

plt.figure(figsize=(6,6))
plt.xlim(0,6)
plt.ylim(0,6)
draw_path(result)
plt.savefig("maze_bfs.png")
plt.show()
````
