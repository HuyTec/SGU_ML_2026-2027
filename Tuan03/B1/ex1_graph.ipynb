# Buổi 4: Tìm kiếm trong không gian trạng thái 

## Bài tập: 
Cho đồ thị `G` = `(V,E)`có 6 đỉnh A, B, C, D, E, F được xác định bởi danh sách cạnh: 
\begin{aligned}
A &\to \{B, C, F\} \\
B &\to \{A, C, E, F\} \\
C &\to \{A, B, D, F\} \\
D &\to \{C, E\} \\
E &\to \{B, D, F\} \\
F &\to \{A, C, E\}
\end{aligned}
Tìm đường đi ngắn nhất từ $A \to D$

### Yêu cầu thêm:
Hãy cho biết bậc của từng đỉnh đồ thị, bậc cao nhất và vẽ đồ thị cho bên trên.

### Input

````python
DSK = {
    'A': {'B', 'C', 'F'},
    'B': {'A', 'C', 'E'},
    'C': {'A', 'B', 'D', 'F'},
    'D': {'C', 'E'},
    'E': {'B', 'D', 'F'},
    'F': {'A', 'C', 'E'}
}
````

### Tạo file dữ liệu

````python
from ex1_graph import make_input
file_path = "dske_ex1.txt"
make_input(DSK, file_path, verbose=True)
````

**dske_ex1.txt**
```
6
A F C B

```

### Đọc dữ liệu


````python
from ex1_graph import load_data
import pprint

graph = load_data(file_path)
pprint.pprint(graph)
````

### Vẽ đồ thị

````python
import matplotlib.pyplot as plt
import networkx as nx

G = nx.Graph()
for u in (graph):
    for v in(graph[u]):
        G.add_edge(u,v)

nodes = sorted(G.nodes()) ## xắp xếp theo thứ tự alphabet
plt.figure(figsize=(8,6))
pos = nx.circular_layout(nodes)
nx.draw(G, pos, with_labels=True, node_size=1000, node_color="pink", width=3)
plt.savefig("dothi2.png")
plt.show()
````
