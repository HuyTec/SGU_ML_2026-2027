**b**. Sinh ngẫu nhiên N diem $(x, y)$ với y là hàm phân phối chuẩn $N(μ = 1, σ^2 = 1.0)$:
$$y = f(x | \mu,\sigma) = \frac{1}{\sigma \sqrt{2\pi}}
\exp\!\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$


nơi mà $x ∈ [μ − 5σ, μ + 5σ]$.

```python
mu = 1.0 ## giá trị trung bình
sigma2 = 1.0 ## phương sai
N = 100
x_min = mu - 5*sigma2
x_max = mu + 5*sigma2
#x = np.linspace(x_min,x_max,N) ## tạo tập hợp 200 điểm rời rạc trong đoạn [1,6]
x = np.random.uniform(x_min, x_max, N)

y = (np.exp(-((x-mu)**2)/(2*sigma2)))/(np.sqrt(sigma2*2*np.pi)) ## Hàm phân phối chuẩn

# Sử dụm hàm np.random.rand() sinh số ngẫu nhiên trong [0,1) --> [0,1) * (b - a) + a --> [a,b)
# Sử dụng hàm sau để tính hàm xác suất chuẩn N(m, s): np.sqrt, np.exp
# ...
print("b. Sinh cặp (x, y) với y làm hàm phân phối chuẩn")
print(f"+ x: {x_min} <= min [{np.min(x): .2f}] <= max [{np.max(x): .2f}] <= {x_max}")
print(f"+ Ve ham so y(x)")
plt.figure(figsize=(6,6)) # kich thuoc ban ve
plt.scatter(x, y) # ve cac diem (xi, yi)
plt.savefig("4b.png") # luu do thi
plt.show() # hien thi do thi
# b. Sinh cặp (x, y) với y làm hàm phân phối chuẩn
# + x: -4.0 <= min [-4.00] <= max [ 5.99] <= 6.0
# + Ve ham so y(x)
```

#### **2.4. Giải hệ phương trình tuyến tính**
$$
\begin{cases}
4x_1 + 3x_2 - 5x_3 = 2 \\
-2x_1 - 4x_2 + 5x_3 = 5 \\
8x_1 + 8x_2 \phantom{- 5x_3} = -3\\
\end{cases}
$$




```python
A = np.array([[4,3,-5],[-2,-4,5],[8, 8, 0]])
y = np.array([2,5,-3])
"""
Từ hệ => ma trận hệ số A và vecto hệ số tự do y : Ax = y <=> x = A^(-1) . y
Khi nhân ma trận dùng @ (not *)
"""

reA = np.linalg.inv(A) ## hàm tính ma trận nghịch đảo
x = reA @ y       
yy = A @ x
diff = np.sqrt(sum((yy - y)**2)) ## Cong thuc tinh chuan L2
# ...
print(f"+ Ma tran A: \n{A}")
print(f"+ Vector y: {y}")
print(f"+ Nghiem x: {x}")
print(f"+ Kiem tra: yy = Ax = {yy} ==> ||yy -y ||_2 = {diff: .2f}")

# + Ma tran A:
# [[ 4 3 -5]
# [-2 -4 5]
# [ 8 8 0]]
# + Vector y: [ 2 5 -3]
# + Nghiem x: [ 2.20833333 -2.58333333 -0.18333333]
# + Kiem tra: yy = Ax = [ 2. 5. -3.] ==> ||yy -y ||_2 = 0.00
```

### **Bài 3: Tính gần đúng số pi dùng phương pháp Monte Carlo**
#### Nội dung:
Vẽ một đường tròn $C$ có bán kính $r$ nội tiếp một hình vuông $C$, và đặt ngẫu nhiên $N$ chấm lên hình vuông.
Tỉ lệ các chấm nằm trong hình tròn trên tổng số chấm xấp xỉ bằng diện tích của hình tròn chia cho hình
vuông như sau:
$$\frac{N_c}{N} \approx \frac{S_c}{S_N}$$
nơi mà $N_C$ là các chấm trong hình tròn, $N$ là tổng số chấm, $S_C$ và $S_N$ lần lượt là diện tích của hình tròn và
hình vuông.
$$S_c = \pi * r^2$$
$$S_N = (2*r)^2 = 4r^2$$
Do đó, công thức trên trở thành:
$$\frac{N_c}{N} \approx \frac{\pi*r^2}{4r^2} = \frac{\pi}{4}$$
Suy ra, ta có giá trị số π được tính như sau:
$$π ≈ \frac{4N_C}{N}$$
Các bạn hãy lập trình tính sấp xỉ số $π$ dùng phương pháp Monte Carlo. Cho biết các sai số với $N = 100$,
$N = 10000$, và $N = 1000000$. Vẽ đồ thị minh họa cho chương trình.

##### Cài đặt hàm tính toán số $\pi$

```python
import numpy as np
def calc_pi_monte_carlo(n=100):
    pi = 0
    # gợi ý: sử dụng np.random.rand(s1, s2, ...) --> sinh ngẫu nhiên các số trong nửa đoạn [0.0, 1.0)
    x =  np.random.rand(n)
    y =  np.random.rand(n)
    
    points = np.column_stack((x,y)) ##tập hợp các Nhóm (x,y) thành một cặp điểm --> vector [[x1,y1],[x2,y2],...,[xn,yn]]
    center = np.array([0,0]) ## tạo mảng với các điểm trung tâm --> vector [[0,0],[0,0]...[0,0]](n)
    dis = np.linalg.norm(points - center, axis=1) ## hàm tính khoảng cách Euclid của vector tao boi diem ngau nhien và tâm
    c_cnt = (dis <= 1).sum() ## đếm số điểm với điều kiện tại đó độ dài vector >=1

    pi = 4*c_cnt/n
    # ...
    return pi
n1, n2, n3 = 100, 10000, 1000000
test_case1 = calc_pi_monte_carlo(n1)
print(f'Voi n = {n1}, pi = {test_case1}')
test_case2 = calc_pi_monte_carlo(n2)
print(f'Voi n = {n2}, pi = {test_case2}')
test_case3 = calc_pi_monte_carlo(n3)
print(f'Voi n = {n3}, pi = {test_case3}')
```

##### Kiểm tra hàm tính toán số $\pi$

```python
import math
print("epsilon(n=100): ", calc_pi_monte_carlo(n = 100) - math.pi)
print("epsilon(n=100): ", calc_pi_monte_carlo(n = 100) - math.pi)
print("epsilon(n=10000): ", calc_pi_monte_carlo(n = 10000) - math.pi)
print("epsilon(n=10000): ", calc_pi_monte_carlo(n = 10000) - math.pi)
print("epsilon(n=1000000): ", calc_pi_monte_carlo(n = 1000000) - math.pi)
print("epsilon(n=1000000): ", calc_pi_monte_carlo(n = 1000000) - math.pi)
# epsilon(n=100): 0.0984073464102071
# epsilon(n=100): -0.061592653589793045
# epsilon(n=10000): -0.04559265358979303
# epsilon(n=10000): -0.009192653589793043
# epsilon(n=1000000): -0.0020086535897929636
# epsilon(n=1000000): -0.0008486535897929137
```
